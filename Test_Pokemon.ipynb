```python
# Create a Pokemon class
class Pokemon():
    def __init__(self, name, level, type):
        self.name = name
        self.level = level
        self.type = type
        self.max_health = level * 10
        self.current_health = self.max_health
        self.knocked_out = False

    def __repr__(self):
        return "Pokemon: {} is level {} and is of type {}. Currently has {} health points".format(self.name, self.level, self.type, self.current_health)

    def lose_health(self, health_lost):
        self.current_health -= health_lost
        if self.current_health <= 0:
            self.current_health = 0
            self.knocked_out = True
            print("{} has been knocked out!".format(self.name))
        else:
            print("{} has lost {} health points and now has {} health points.".format(self.name, health_lost, self.current_health))

    def gain_health(self, health_gained):
        if self.knocked_out == True:
            print("{} is knocked out and cannot gain health.".format(self.name))
        else:
            self.current_health += health_gained
            if self.current_health > self.max_health:
                self.current_health = self.max_health
            print("{} has gained {} health points and now has {} health points.".format(self.name, health_gained, self.current_health))

    def knock_out(self):
        self.knocked_out = True
        self.current_health = 0
        print("{} has been knocked out!".format(self.name))

    def revive(self):
        self.knocked_out = False
        self.current_health = self.max_health
        print("{} has been revived and now has {} health points.".format(self.name, self.current_health))
```

```python
Pikachu = Pokemon("Pikachu", 10, "Electric")
Charmander = Pokemon("Charmander", 10, "Fire")
Squirtle = Pokemon("Squirtle", 10, "Water")
Bulbasaur = Pokemon("Bulbasaur", 10, "Grass")
print(Pikachu)
```

```python
Pikachu.lose_health(100)
```

```python
Pikachu.revive()
```
